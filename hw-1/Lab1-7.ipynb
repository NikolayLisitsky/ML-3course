8. Выведите минимальное и максимальное представимое значение для каждого скалярного типа numpy

````python
types = [np.bool_, np.int8, np.int16, np.int32, np.int64,
         np.uint8, np.uint16, np.uint32, np.uint64,
         np.float16, np.float32, np.float64,
         np.complex64, np.complex128]
for dtype in types:
    if dtype is np.bool_:
        print(f'{dtype.__name__:12} min=False, max=True')
    elif np.issubdtype(dtype, np.integer):
        info = np.iinfo(dtype)
        print(f'{dtype.__name__:12} min={info.min}, max={info.max}')
    else:
        real_dtype = np.empty((), dtype=dtype).real.dtype
        info = np.finfo(real_dtype)
        if np.issubdtype(dtype, np.complexfloating):
            print(f'{dtype.__name__:12} |Re|, |Im| <= {info.max:.5g} (диапазон каждой компоненты)')
        else:
            print(f'{dtype.__name__:12} min={-info.max:.5g}, max={info.max:.5g}, ' f'наименьшее положительное normal={info.tiny:.5g}')
````

````python
import pandas as pd
````

````python
from pathlib import Path
from IPython.display import display
path = Path('data.csv')
if not path.exists():
    path = Path('Variant 7/data.csv')
df = pd.read_csv(path)
print('Размер датасета:', df.shape)
display(df.head())
df[['Systolic BP', 'Diastolic BP']] = df['Blood Pressure (s/d)'].str.split('/', expand=True).astype(float)
````

1. Какой средний вес людей представленных в наборе данных?

````python
mean_weight = df['Weight (kg)'].mean()
print(f'Средний вес: {mean_weight:.2f} кг')
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
counts = df['Gender'].value_counts()
print(counts.to_string())
print((counts / len(df) * 100).round(2).astype(str) + '%')
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
q25, q75 = df['Stress Levels'].quantile([0.25, 0.75])
low = df[df['Stress Levels'] <= q25]
high = df[df['Stress Levels'] >= q75]
for label, group in [('Низкий стресс (нижние 25%)', low), ('Высокий стресс (верхние 25%)', high)]:
    print(f'{label}: n={len(group)}, САД={group["Systolic BP"].mean():.2f}, ' f'ДАД={group["Diastolic BP"].mean():.2f} мм рт. ст.')
print('Давление выше при сильном стрессе:',  'да' if high['Systolic BP'].mean() > low['Systolic BP'].mean() and high['Diastolic BP'].mean() > low['Diastolic BP'].mean() else 'не для обеих компонент')
print('Это описательная связь, а не доказательство причинности.')
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
grouped = df.groupby('Gender')['Cholesterol Level (mg/dL)'].agg(['count', 'mean', 'median', 'std'])
display(grouped.round(2))
means = grouped['mean']
print(f'Разность средних (мужчины − женщины): {means["Male"] - means["Female"]:.2f} мг/дл')
from scipy.stats import ttest_ind
male = df.loc[df['Gender']=='Male', 'Cholesterol Level (mg/dL)'].dropna()
female = df.loc[df['Gender']=='Female', 'Cholesterol Level (mg/dL)'].dropna()
pvalue = ttest_ind(male, female, equal_var=False).pvalue
print(f'Двусторонний t-тест Уэлча: p={pvalue:.4g}; статистически значимое отличие при α=0.05:', pvalue < 0.05)
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df['Gender'] == 'Female']
young = women[women['Age (years)'] <= 30]['Cognitive Function']
older = women[women['Age (years)'] >= 60]['Cognitive Function']
print(f'Женщины до 30 включительно: n={len(young)}, средняя оценка={young.mean():.2f}')
print(f'Женщины 60+: n={len(older)}, средняя оценка={older.mean():.2f}')
print('У женщин 60+ оценка ниже:', older.mean() < young.mean())
print('Сравнение возрастных групп не доказывает изменение у одного и того же человека.')
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
over50 = df[df['Age (years)'] > 50]
activity_weights = over50.groupby('Physical Activity Level')['Weight (kg)'].agg(['count', 'mean', 'median'])
display(activity_weights.reindex(['Low','Moderate','High']).round(2))
print('Разность средних весов Low − High:', round(activity_weights.loc['Low','mean'] - activity_weights.loc['High','mean'], 2), 'кг')
print('Эта разность показывает ассоциацию, а не доказанное влияние активности на вес.')
````

7. Сравните минимальный вес мужчин с высоким и низким заработком

````python
men = df[df['Gender'] == 'Male']
mins = men[men['Income Level'].isin(['Low', 'High'])].groupby('Income Level')['Weight (kg)'].min()
print(mins.round(2).to_string())
print(f'Разница High − Low: {mins["High"]-mins["Low"]:.2f} кг')
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
p30, p60, p75 = df['BMI'].quantile([.30, .60, .75])
above75 = df[df['BMI'] > p75]
between = df[df['BMI'].between(p30, p60, inclusive='both')]
for label, group in [('BMI > 75-го перцентиля', above75), ('BMI от 30-го до 60-го перцентиля', between)]:
    disease = group['Chronic Diseases'].notna().mean()
    print(f'{label}: n={len(group)}, есть заболевание у {disease:.2%}')
print('Чаще в группе BMI > 75-го перцентиля:', above75['Chronic Diseases'].notna().mean() > between['Chronic Diseases'].notna().mean())
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
income_activity = pd.crosstab(df['Income Level'], df['Physical Activity Level'], normalize='index') * 100
display(income_activity.reindex(index=['Low','Medium','High'], columns=['Low','Moderate','High']).round(2))
print('Высокая активность чаще при высоком доходе:', income_activity.loc['High','High'] > income_activity.loc['Low','High'])
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
men = df[df['Gender'] == 'Male']
threshold = df['Sun Exposure'].quantile(.75)  
men_high_sun = men[men['Sun Exposure'] > threshold]
share = men_high_sun['Income Level'].eq('High').mean()
print(f'Порог высокого пребывания на солнце (> 75-го перцентиля): {threshold:.2f} ч/нед')
print(f'Мужчин с высоким пребыванием на солнце: {len(men_high_sun)}')
print(f'Из них с высоким доходом: {men_high_sun["Income Level"].eq("High").sum()}')
print(f'Искомая доля: {share:.2%}')
````

2. Как связана оценка собственных когнитивных функций с уровнем стресса?

````python
r = df['Cognitive Function'].corr(df['Stress Levels'])
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['Stress Levels'], df['Cognitive Function'], s=13, alpha=.22)
coef = np.polyfit(df['Stress Levels'], df['Cognitive Function'], 1)
xline = np.linspace(df['Stress Levels'].min(), df['Stress Levels'].max(), 100)
ax.plot(xline, np.polyval(coef, xline))
ax.set(title=f'Когнитивная оценка и стресс: корреляция r={r:.3f}',xlabel='Уровень стресса', ylabel='Когнитивная функция')
plt.tight_layout(); plt.show()
print(f'Корреляция Пирсона: {r:.3f}.','Слабая линейная связь.' if abs(r)<0.3 else 'Заметная линейная связь.')
````
